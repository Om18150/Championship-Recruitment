```python
import pandas as pd
import numpy as np

main = pd.read_csv('players.csv')
midstats = pd.read_csv('miscplay.csv')
shooting = pd.read_csv('shooting.csv')
merged = pd.merge(main, midstats, on=["Rk", "Player", "Nation", "Pos", "Squad", "Comp", "Age", "Born", "Matches", "-9999", "CrdY", "CrdR", "90s"], how='left')
merged = pd.merge(merged, shooting, on=["Rk", "Player", "Nation", "Pos", "Squad", "Comp", "Age", "Born", "Matches", "-9999", "Gls","PK", "PKatt","90s"], how='left')
merged = merged.drop(columns = ["Gls.1", "Ast.1", "G+A.1", "G-PK.1", "Rk", "-9999"])
merged.head()
merged.to_csv('merged.csv', index=False)

```

Merging all different player stat tables and cleaning data.

```python
main = main[main["Min"] >= 900]
main = main[main["Age"] <= 23]
main = main[main["Pos"] == "MF"]
main.info()
```

```python
merged = merged[merged["Min"] >= 900]
merged = merged[merged["Age"] <= 23]
merged = merged[merged["Pos"] == "MF"]
merged.info()
merged.head()
```

merged.info() and main.info() check that no player has been lost in the merging process since both show 172 entries.
